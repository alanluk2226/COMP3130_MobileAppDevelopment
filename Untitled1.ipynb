```python
import numpy as np
from tensorflow.keras.datasets import cifar100

ANIMALS_PEOPLE_INDICES = [
    1, 2, 3, 4, 6, 7, 11, 14, 15, 18, 19, 21, 24, 26, 27, 29, 30, 31, 32, 34, 35,
    36, 38, 42, 43, 44, 45, 46, 50, 55, 57, 63, 64, 65, 66, 67, 72, 73, 74, 75,
    77, 78, 79, 80, 91, 93, 95, 97, 98, 99
]

def select_animals_people(y):
    return np.isin(np.ravel(y), ANIMALS_PEOPLE_INDICES)

def compress_animals_people_indices(y):
    out = np.zeros(100)
    for i, index in enumerate(ANIMALS_PEOPLE_INDICES):
        out[index] = i
    return out[y]

def load_cifar100_animals_people():
    (X_train, y_train), (X_test, y_test) = cifar100.load_data()
    mask_train = select_animals_people(y_train)
    X_train, y_train = X_train[mask_train], y_train[mask_train]
    mask_test = select_animals_people(y_test)
    X_test, y_test = X_test[mask_test], y_test[mask_test]
    y_train = compress_animals_people_indices(y_train)
    y_test = compress_animals_people_indices(y_test)
    return X_train, y_train, X_test, y_test

X_train, y_train, X_test, y_test = load_cifar100_animals_people()
print(f"Training data shape: {X_train.shape}")
```

```python
from sklearn.model_selection import train_test_split


X_train_sub, X_val, y_train_sub, y_val = train_test_split(
    X_train, y_train, test_size=0.2, random_state=42
)

IMG_SIZE = 128

print("Data split successfully!")
```

```python
import tensorflow as tf
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input
from tensorflow.keras import models, layers


X_train_resized = tf.image.resize(X_train_sub, (IMG_SIZE, IMG_SIZE))
X_val_resized = tf.image.resize(X_val, (IMG_SIZE, IMG_SIZE))

X_train_resnet = preprocess_input(X_train_resized)
X_val_resnet = preprocess_input(X_val_resized)


base_resnet = ResNet50(input_shape=(IMG_SIZE, IMG_SIZE, 3), include_top=False, weights='imagenet')
base_resnet.trainable = False

model2 = models.Sequential([
    base_resnet,
    layers.GlobalAveragePooling2D(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(50, activation='softmax')
])

model2.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])


datagen = tf.keras.preprocessing.image.ImageDataGenerator(rotation_range=15, horizontal_flip=True)

print("Starting training for Model 2 (ResNet50)...")
history2 = model2.fit(
    datagen.flow(X_train_resnet, y_train_sub, batch_size=32),
    validation_data=(X_val_resnet, y_val),
    epochs=5
)
print("Training Complete!")
```

```python
import tensorflow as tf
from sklearn.model_selection import train_test_split
from tensorflow.keras.applications import ResNet50
from tensorflow.keras import models, layers

X_train_sub, X_val, y_train_sub, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)

IMG_SIZE = 128
BATCH_SIZE = 32

def preprocess(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.keras.applications.resnet50.preprocess_input(image)
    return image, label

train_ds = tf.data.Dataset.from_tensor_slices((X_train_sub, y_train_sub))
train_ds = train_ds.shuffle(1000).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

val_ds = tf.data.Dataset.from_tensor_slices((X_val, y_val))
val_ds = val_ds.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)


base_resnet = ResNet50(input_shape=(IMG_SIZE, IMG_SIZE, 3), include_top=False, weights='imagenet')
base_resnet.trainable = False

model2 = models.Sequential([
    base_resnet,
    layers.GlobalAveragePooling2D(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(50, activation='softmax')
])
model2.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])


print("Starting training for Model 2 (ResNet50)...")
history2 = model2.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)
print("Training Complete!")
```

```python
import tensorflow as tf
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras import models, layers

BATCH_SIZE = 32


def preprocess_eff(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.keras.applications.efficientnet.preprocess_input(image)
    return image, label


train_ds_eff = tf.data.Dataset.from_tensor_slices((X_train_sub, y_train_sub))
train_ds_eff = train_ds_eff.shuffle(1000).map(preprocess_eff, num_parallel_calls=tf.data.AUTOTUNE).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

val_ds_eff = tf.data.Dataset.from_tensor_slices((X_val, y_val))
val_ds_eff = val_ds_eff.map(preprocess_eff, num_parallel_calls=tf.data.AUTOTUNE).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

base_eff = EfficientNetB0(input_shape=(IMG_SIZE, IMG_SIZE, 3), include_top=False, weights='imagenet')
base_eff.trainable = False

model3 = models.Sequential([
    base_eff,
    layers.GlobalAveragePooling2D(),
    layers.Dense(256, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(50, activation='softmax')
])

model3.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])


print("Starting training for Model 3 (EfficientNetB0)...")
history3 = model3.fit(
    train_ds_eff,
    validation_data=val_ds_eff,
    epochs=5
)
print("Training Complete!")
```

```python
import matplotlib.pyplot as plt


loss2, acc2 = model2.evaluate(val_ds, verbose=0)
loss3, acc3 = model3.evaluate(val_ds_eff, verbose=0)

print("--- Final Validation Results ---")
print(f"Model 2 (ResNet50) Val Accuracy: {acc2*100:.2f}%")
print(f"Model 3 (EfficientNet) Val Accuracy: {acc3*100:.2f}%")


plt.figure(figsize=(10, 5))
plt.plot(history2.history['val_accuracy'], label='Model 2 (ResNet50)', color='blue', linewidth=2)
plt.plot(history3.history['val_accuracy'], label='Model 3 (EfficientNet)', color='green', linewidth=2)
plt.title('Validation Accuracy: Model 2 vs Model 3')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```
